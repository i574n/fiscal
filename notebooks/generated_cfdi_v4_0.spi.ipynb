```spiral
///- --test
///> rust

inl example = x_comprobante_example 0i32
inl document =
    xml.element
        "Comprobante"
        (("xmlns", "http://www.sat.gob.mx/cfd/4") :: x_comprobante_attributes example)
        (x_comprobante_children example)
file_system.directory_create_all (generated_fiscal_runtime.sample_path "") |> ignore
file_system.write_all_text
    (generated_fiscal_runtime.sample_path "generated_cfdi_v4_0_example.xml")
    (xml.write_node (xml.with_namespace "Signature" "http://www.w3.org/2000/09/xmldsig#" document))
inl documents = [ "deps/corpus/mx/stamped/cfdiutils-cfdi40-valid.xml" ]
inl check (path : string) : list string =
    inl text = file_system.read_all_text (generated_fiscal_runtime.sample_path (sm'.(++#) "../" path))
    match xml.read_xml text with
    | node :: [] =>
        match xml_codec.read_back_difference (x_comprobante_of_xml) (x_comprobante_to_xml "Comprobante") node with
        | Some difference => [ sm'.(++#) path (sm'.(++#) ": " difference) ]
        | None => []
    | _ => [ sm'.(++#) path ": not one root element" ]
inl failures = listm.concat (listm.map check documents)
inl _ =
    match failures with
    | first :: _ =>
        inl count = sm'.obj_to_string (listm.length failures : i32)
        inl message = sm'.(++#) " real documents do not read back unchanged; first: " first
        failwith (sm'.(++#) count message)
    | [] => ()
inl text = file_system.read_all_text (generated_fiscal_runtime.sample_path "generated_cfdi_v4_0_example.xml")
let rec runs (left : i32) (parses : list i64) (writes : list i64) (bytes : i64) : list i64 * list i64 * i64 =
    if left = 0i32
    then parses, writes, bytes
    else
        inl watch = date_time.stopwatch ()
        date_time.stopwatch_start watch
        inl start = date_time.stopwatch_elapsed_microseconds watch
        match xml.read_xml text with
        | node :: [] =>
            inl parsed = xml_codec.expect_decoded (x_comprobante_of_xml node)
            inl parsed_at = date_time.stopwatch_elapsed_microseconds watch
            inl written = xml.write_node (x_comprobante_to_xml "Comprobante" parsed)
            inl written_at = date_time.stopwatch_elapsed_microseconds watch
            inl parses = generated_fiscal_runtime.insert_run (parsed_at - start) parses
            inl writes = generated_fiscal_runtime.insert_run (written_at - parsed_at) writes
            runs (left - 1i32) parses writes (bytes + sm'.length written)
        | _ => failwith "one Comprobante element"
inl _ = runs (dyn 5i32) [] [] 0i64
inl parses, writes, bytes = runs (dyn 31i32) [] [] 0i64
inl row = sm'.(++#) "generated_cfdi_v4_0 parse_us " (generated_fiscal_runtime.median parses)
inl row = sm'.(++#) row (sm'.(++#) " write_us " (generated_fiscal_runtime.median writes))
inl row = sm'.(++#) row (sm'.(++#) " runs 31 bytes " (sm'.obj_to_string bytes))
console.write_line row
match xml.read_xml text with
| node :: [] =>
    inl parsed = xml_codec.expect_decoded (x_comprobante_of_xml node)
    inl document =
        xml.element
            "Comprobante"
            (("xmlns", "http://www.sat.gob.mx/cfd/4") :: x_comprobante_attributes parsed)
            (x_comprobante_children parsed)
    inl folder = generated_fiscal_runtime.sample_path "../target/bench"
    match file_system.directory_create_all folder with
    | Ok _ =>
        file_system.write_all_text
            (sm'.(++#) folder "/ours-generated_cfdi_v4_0_example.xml")
            (xml.write_node (xml.with_namespace "Signature" "http://www.w3.org/2000/09/xmldsig#" document))
    | Error _ => failwith "cannot create target/bench"
| _ => failwith "one Comprobante element"
```
