```spiral
///- --package . --timeout 1200000
:< sm'_operators

:! gravado = ()
:! exonerado = ()
:! inafecto = ()

:| igv_affectation g =
    | GravadoOperacionOnerosa :: igv_affectation gravado
    | GravadoRetiroPorPremio :: igv_affectation gravado
    | GravadoRetiroPorDonacion :: igv_affectation gravado
    | GravadoRetiro :: igv_affectation gravado
    | GravadoRetiroPorPublicidad :: igv_affectation gravado
    | GravadoBonificaciones :: igv_affectation gravado
    | GravadoRetiroPorEntregaATrabajadores :: igv_affectation gravado
    | GravadoIvap :: igv_affectation gravado
    | ExoneradoOperacionOnerosa :: igv_affectation exonerado
    | ExoneradoTransferenciaGratuita :: igv_affectation exonerado
    | Exportacion :: igv_affectation exonerado
    | InafectoOperacionOnerosa :: igv_affectation inafecto
    | InafectoRetiroPorBonificacion :: igv_affectation inafecto
    | InafectoRetiro :: igv_affectation inafecto
    | InafectoRetiroPorMuestrasMedicas :: igv_affectation inafecto
    | InafectoRetiroPorConvenioColectivo :: igv_affectation inafecto
    | InafectoRetiroPorPremio :: igv_affectation inafecto
    | InafectoRetiroPorPublicidad :: igv_affectation inafecto

:| affectation =
    | Taxed : igv_affectation gravado
    | Exempt : igv_affectation exonerado
    | Unaffected : igv_affectation inafecto

:= affectation_code \* g. (a : igv_affectation g) : string =
    ?= a ?|
    | GravadoOperacionOnerosa => "10"
    | GravadoRetiroPorPremio => "11"
    | GravadoRetiroPorDonacion => "12"
    | GravadoRetiro => "13"
    | GravadoRetiroPorPublicidad => "14"
    | GravadoBonificaciones => "15"
    | GravadoRetiroPorEntregaATrabajadores => "16"
    | GravadoIvap => "17"
    | ExoneradoOperacionOnerosa => "20"
    | ExoneradoTransferenciaGratuita => "21"
    | Exportacion => "40"
    | InafectoOperacionOnerosa => "30"
    | InafectoRetiroPorBonificacion => "31"
    | InafectoRetiro => "32"
    | InafectoRetiroPorMuestrasMedicas => "33"
    | InafectoRetiroPorConvenioColectivo => "34"
    | InafectoRetiroPorPremio => "35"
    | InafectoRetiroPorPublicidad => "36"

:= taxed_cases () : list (igv_affectation gravado) =
    GravadoOperacionOnerosa
    :: GravadoRetiroPorPremio
    :: GravadoRetiroPorDonacion
    :: GravadoRetiro
    :: GravadoRetiroPorPublicidad
    :: GravadoBonificaciones
    :: GravadoRetiroPorEntregaATrabajadores
    :: GravadoIvap
    :: []

:= exempt_cases () : list (igv_affectation exonerado) =
    ExoneradoOperacionOnerosa :: ExoneradoTransferenciaGratuita :: Exportacion :: []

:= unaffected_cases () : list (igv_affectation inafecto) =
    InafectoOperacionOnerosa
    :: InafectoRetiroPorBonificacion
    :: InafectoRetiro
    :: InafectoRetiroPorMuestrasMedicas
    :: InafectoRetiroPorConvenioColectivo
    :: InafectoRetiroPorPremio
    :: InafectoRetiroPorPublicidad
    :: []

:= taxed_of (code : string) : option (igv_affectation gravado) =
    listm'::try_find_ (\ a => affectation_code a = code) (taxed_cases ())

:= exempt_of (code : string) : option (igv_affectation exonerado) =
    listm'::try_find_ (\ a => affectation_code a = code) (exempt_cases ())

:= unaffected_of (code : string) : option (igv_affectation inafecto) =
    listm'::try_find_ (\ a => affectation_code a = code) (unaffected_cases ())

:= affectation_of (code : string) : option affectation =
    ?= taxed_of code ?|
    | Some a => Some (Taxed a)
    | None =>
        ?= exempt_of code ?|
        | Some a => Some (Exempt a)
        | None => optionm::map (\ a => Unaffected a) (unaffected_of code)

:= code_of (a : affectation) : string =
    ?= a ?|
    | Taxed g => affectation_code g
    | Exempt e => affectation_code e
    | Unaffected u => affectation_code u

:| document_type =
    | Factura
    | Boleta
    | NotaCredito
    | NotaDebito
    | GuiaRemisionRemitente

:= document_type_code (d : document_type) : string =
    ?= d ?|
    | Factura => "01"
    | Boleta => "03"
    | NotaCredito => "07"
    | NotaDebito => "08"
    | GuiaRemisionRemitente => "09"

:| credit_note_reason =
    | AnulacionDeLaOperacion
    | AnulacionPorErrorEnElRuc
    | CorreccionPorErrorEnLaDescripcion
    | DescuentoGlobal
    | DescuentoPorItem
    | DevolucionTotal
    | DevolucionPorItem
    | Bonificacion
    | DisminucionEnElValor
    | OtrosConceptos

:= credit_note_reason_code (r : credit_note_reason) : string =
    ?= r ?|
    | AnulacionDeLaOperacion => "01"
    | AnulacionPorErrorEnElRuc => "02"
    | CorreccionPorErrorEnLaDescripcion => "03"
    | DescuentoGlobal => "04"
    | DescuentoPorItem => "05"
    | DevolucionTotal => "06"
    | DevolucionPorItem => "07"
    | Bonificacion => "08"
    | DisminucionEnElValor => "09"
    | OtrosConceptos => "10"

:| debit_note_reason =
    | InteresPorMora
    | AumentoEnElValor
    | PenalidadOtrosConceptos

:= debit_note_reason_code (r : debit_note_reason) : string =
    ?= r ?|
    | InteresPorMora => "01"
    | AumentoEnElValor => "02"
    | PenalidadOtrosConceptos => "03"

:| price_type =
    | PrecioUnitarioIncluyeIgv
    | ValorReferencialUnitarioEnOperacionesNoOnerosas

:= price_type_code (p : price_type) : string =
    ?= p ?|
    | PrecioUnitarioIncluyeIgv => "01"
    | ValorReferencialUnitarioEnOperacionesNoOnerosas => "02"

:~ ^ coded \* t. (code : string) (encode : t -> string) (cases : list t) : option t =
    ?= cases ?|
    | c :: rest => ? encode c = code ?> Some c |: coded code encode rest
    | [] => None

:= all_document_types () : list document_type = [ Factura; Boleta; NotaCredito; NotaDebito; GuiaRemisionRemitente ]

:= all_credit_note_reasons () : list credit_note_reason =
    [
        AnulacionDeLaOperacion
        AnulacionPorErrorEnElRuc
        CorreccionPorErrorEnLaDescripcion
        DescuentoGlobal
        DescuentoPorItem
        DevolucionTotal
        DevolucionPorItem
        Bonificacion
        DisminucionEnElValor
        OtrosConceptos
    ]

:= all_debit_note_reasons () : list debit_note_reason = [ InteresPorMora; AumentoEnElValor; PenalidadOtrosConceptos ]

:= document_type_of (code : string) : option document_type = coded code document_type_code (all_document_types ())

:= price_type_of (code : string) : option price_type =
    coded code price_type_code [ PrecioUnitarioIncluyeIgv; ValorReferencialUnitarioEnOperacionesNoOnerosas ]

:= credit_note_reason_of (code : string) : option credit_note_reason =
    coded code credit_note_reason_code (all_credit_note_reasons ())

:= debit_note_reason_of (code : string) : option debit_note_reason =
    coded code debit_note_reason_code (all_debit_note_reasons ())

:| tax_scheme =
    | Igv
    | ImpuestoArrozPilado
    | Isc
    | ExportacionScheme
    | Gratuito
    | ExoneradoScheme
    | InafectoScheme
    | IcbPer
    | Otros

:= tax_scheme_code (s : tax_scheme) : string =
    ?= s ?|
    | Igv => "1000"
    | ImpuestoArrozPilado => "1016"
    | Isc => "2000"
    | ExportacionScheme => "9995"
    | Gratuito => "9996"
    | ExoneradoScheme => "9997"
    | InafectoScheme => "9998"
    | IcbPer => "7152"
    | Otros => "9999"

:= all_tax_schemes () : list tax_scheme =
    [ Igv; ImpuestoArrozPilado; Isc; ExportacionScheme; Gratuito; ExoneradoScheme; InafectoScheme; IcbPer; Otros ]

:= tax_scheme_of (code : string) : option tax_scheme = coded code tax_scheme_code (all_tax_schemes ())

:= is_onerosa \* g. (a : igv_affectation g) : bool =
    ?= a ?|
    | GravadoOperacionOnerosa => !+
    | GravadoIvap => !+
    | ExoneradoOperacionOnerosa => !+
    | InafectoOperacionOnerosa => !+
    | Exportacion => !+
    | _ => !-

:= tax_category \* g. (a : igv_affectation g) : tax_scheme =
    ?= a ?|
    | GravadoOperacionOnerosa => Igv
    | GravadoIvap => ImpuestoArrozPilado
    | ExoneradoOperacionOnerosa => ExoneradoScheme
    | InafectoOperacionOnerosa => InafectoScheme
    | Exportacion => ExportacionScheme
    | _ => Gratuito

:= affectation_scheme (a : affectation) : tax_scheme =
    ?= a ?|
    | Taxed g => tax_category g
    | Exempt e => tax_category e
    | Unaffected u => tax_category u

:= scheme_code_of (a : affectation) : string = tax_scheme_code (affectation_scheme a)

:~ ^ child_named (name : string) (nodes : list xml::xml_node) : option xml::xml_node =
    ?= nodes ?|
    | node :: rest => ? xml::local_name (xml::node_name node) = name ?> Some node |: child_named name rest
    | [] => None

:= child_text (name : string) (node : xml::xml_node) : option string =
    := text_of = \ c => sm'::trim (xml::text_in (xml::element_children c))
    optionm::map text_of (child_named name (xml::element_children node))

:= scheme_verdict (reason : string) (found : string) : option string =
    ?= affectation_of reason ?|
    | None => Some ("TaxExemptionReasonCode " ++# reason ++# " is not in catalog 07")
    | Some a ?& scheme_code_of a = found => None
    | Some a => Some ("reason " ++# reason ++# " expects TaxScheme " ++# scheme_code_of a ++# ", found " ++# found)

:= scheme_finding (reason : string) (category : xml::xml_node) : option string =
    ?= child_named "TaxScheme" (xml::element_children category) ?|
    | None => Some ("TaxExemptionReasonCode " ++# reason ++# " without a TaxScheme")
    | Some scheme => scheme_verdict reason (?= child_text "ID" scheme ?| Some id => id | None => "")

:! category_check = i32 * list string

:= tallied (finding : option string) (category_check (checked, findings)) : category_check =
    ?= finding ?|
    | Some f => category_check (checked + 1i32, f :: findings)
    | None => category_check (checked + 1i32, findings)

:= category_here (node : xml::xml_node) (acc : category_check) : category_check =
    ? xml::local_name (xml::node_name node) <> "TaxCategory"
    ?> acc
    |:
        ?= child_text "TaxExemptionReasonCode" node ?|
        | None => acc
        | Some reason => tallied (scheme_finding reason node) acc

:~ ^ category_findings (node : xml::xml_node) (acc : category_check) : category_check =
    findings_within (xml::element_children node) (category_here node acc)

:& :~ findings_within (nodes : list xml::xml_node) (acc : category_check) : category_check =
    ?= nodes ?|
    | node :: rest => findings_within rest (category_findings node acc)
    | [] => acc

:= affectation_is_onerosa (a : affectation) : bool =
    ?= a ?|
    | Taxed g => is_onerosa g
    | Exempt e => is_onerosa e
    | Unaffected u => is_onerosa u

:= expected_price_type (a : affectation) : price_type =
    ? affectation_is_onerosa a ?> PrecioUnitarioIncluyeIgv |: ValorReferencialUnitarioEnOperacionesNoOnerosas

:~ ^ descendant_text (name : string) (node : xml::xml_node) : option string =
    ?= child_text name node ?|
    | Some text => Some text
    | None => descendant_text_within name (xml::element_children node)

:& :~ descendant_text_within (name : string) (nodes : list xml::xml_node) : option string =
    ?= nodes ?|
    | node :: rest =>
        ?= descendant_text name node ?|
        | Some text => Some text
        | None => descendant_text_within name rest
    | [] => None

:= price_verdict (reason : string) (found : string) : option string =
    ?= optionm::map (\ a => price_type_code (expected_price_type a)) (affectation_of reason) ?|
    | Some expected ?& expected <> found =>
        Some ("reason " ++# reason ++# " expects PriceTypeCode " ++# expected ++# ", found " ++# found)
    | _ => None

:= price_finding (line : xml::xml_node) : option string =
    ?= descendant_text "TaxExemptionReasonCode" line ?|
    | None => None
    | Some reason => price_verdict reason (?= descendant_text "PriceTypeCode" line ?| Some p => p | None => "")

:= is_line (node : xml::xml_node) : bool =
    := local = xml::local_name (xml::node_name node)
    listm'::contains local [ "InvoiceLine"; "CreditNoteLine"; "DebitNoteLine" ]

:~ ^ line_findings (node : xml::xml_node) (acc : category_check) : category_check =
    ? is_line node ?> tallied (price_finding node) acc |: lines_within (xml::element_children node) acc

:& :~ lines_within (nodes : list xml::xml_node) (acc : category_check) : category_check =
    ?= nodes ?|
    | node :: rest => lines_within rest (line_findings node acc)
    | [] => acc

:= pe_entries (manifest : string) : list string =
    := text = xml::read_text_replacing_invalid_utf8 (fiscal_paths::under manifest)
    listm'::filter (\ (line : string) => sm'::starts_with "pe/" line) (listm::map sm'::trim (sm'::split_list "\n" text))

:= homologation_entries () : list string =
    listm::concat [
        pe_entries "deps/corpus/ubl-invoice-documents.txt"
        pe_entries "deps/corpus/ubl-creditnote-documents.txt"
        pe_entries "deps/corpus/ubl-debitnote-documents.txt"
    ]

:= document_check (acc : category_check) (entry : string) : category_check =
    := text = xml::read_text_replacing_invalid_utf8 (fiscal_paths::under ("deps/corpus/" ++# entry))
    findings_within (xml::read_xml text) acc

:= line_check (acc : category_check) (entry : string) : category_check =
    := text = xml::read_text_replacing_invalid_utf8 (fiscal_paths::under ("deps/corpus/" ++# entry))
    lines_within (xml::read_xml text) acc

:= checked_of (category_check (checked, _)) : i32 = checked

:= findings_of (category_check (_, findings)) : list string = listm::rev findings

:= igv_line (rate_percent : i32) (line : igv_affectation gravado) (base_cents : i64) : i64 =
    base_cents * (conv rate_percent : i64) / 100i64
```
